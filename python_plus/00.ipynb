요구사항

 

다음 학생 데이터가 있습니다.

민수 / 85
지영 / 72
현우 / 91

학생을 한 명씩 확인하고,

• 80점 이상이면 합격
• 그렇지 않으면 불합격

을 출력하세요.

```python
students = [
    {"name": "민수", "score": 85},
    {"name": "지영", "score": 72},
    {"name": "현우", "score": 91},
]

for student in students:
    name = student["name"]
    score = student["score"]
    if score >= 80:
        result = "합격"
    else:
        result = "불합격"   

    print(name, score, result)
```

실습 1-2. 쇼핑몰 주문

문제

여러 고객의 이름과 주문 금액이 주어져 있습니다.
고객을 한 명씩 확인하여 주문 금액이 50,000원 이상이면 무료배송, 그렇지 않으면 배송비 있음이라고 표시합니다.


```python
orders = [
    {"customers": "진수", "amount": 600000},
    {"customers": "미진", "amount": 380000},
    {"customers": "아린", "amount": 90000},
]

for order in orders:
    customers = order["customers"]
    amount = order["amount"]

    if amount >= 50000:
        result = "무료배송"
    else:
        result = "배송비 있음"

print(customers, amount, result)
```

종합 실습 1 — 학생 합격 판정

```python
students = [
    {"name": "민수", "score": 85},
    {"name": "지영", "score": 72},
    {"name": "현우", "score": 91},
    {"name": "수진", "score": 80},
]

for student in students:
    if student["score"] >= 80:
        result = "합격"
    else:
        result = "불합격"
    print(student["name"], student["score"], result)
```

쇼핑몰 무료배송 판정

```python
customers = [
    {"customer": "민수", "amount": 45000},
    {"customer": "지영", "amount": 72000},
    {"customer": "현우", "amount": 53000},
    {"customer": "수진", "amount": 49000},
]

for customer in customers:
    name = customer["customer"]
    amount = customer["amount"]

    if amount >= 50000:
        result = "무료배송"
    else:
        result = "배송비 있음"

    print(name, amount, result)
```

상품 재고 상태

 출석률 확인

```python
students = [
    {"name": "민수", "rate": 92},
    {"name": "지영", "rate": 74},
    {"name": "현우", "rate": 88},
    {"name": "수진", "rate": 79},
]

for student in students:
    name = student["name"]
    rate = student["rate"]

    if rate >= 80:
        result = "출석 충족"
    else:
        result = "출석 부족"

    print(name, rate, result)
```

기온 경고

```python
cities = [
    {"city": "서울", "temp": 34},
    {"city": "부산", "temp": 29},
    {"city": "대전", "temp": 33},
    {"city": "광주", "temp": 31},
]

for city in cities:
    name = city["city"]
    temp = city["temp"]

    if temp >= 33:
        result = "폭염 주의"
    else:
        result = "일반"

    print(name, temp, result)
```

회원 구매액 판정

```python
members = [
    {"name": "민수", "amount": 120000},
    {"name": "지영", "amount": 85000},
    {"name": "현우", "amount": 210000},
    {"name": "수진", "amount": 99000},
]

for member in members:
    name = member["name"]
    amount = member["amount"]

    if amount >= 100000:
        grade = "우수회원"
    else:
        grade = "일반회원"

    print(name, amount, grade)
```
